```python
import pandas as pd
import numpy as np

data = {
    'name':['thrisha','vaish','vachana','amit','bebo','divya','asha','arun','putta','ram','radha'],
    'age':[28,25,45,68.15,47,26,27,47,35,14,np.nan], # Added np.nan to make the length 11
    'marks':[90,35,57,90,np.nan,35,67,37,98,34,68],
    'city':['mysuru','hassan','dubai','uganda','mysuru','hassan','dubai','uganda','assam','kodagu','andra']

}

df=pd.DataFrame(data)
print("dataset:")
print(df)
```

```python
print("\nShape of dataset (rows,columns):",df.shape)
```

```python
print("\nColumn names:",df.columns.tolist())
```

```python
print("\nData types of each column:")
print(df.dtypes)
```

```python
print("\ndataset info:")
df.info()
```

```python
print("\Descriptive statistics (numerical columns):")
print(df.describe())
```

```python
print("\nMean:\n",df.mean(numeric_only=True))
```

```python
print("\nMedian:\n",df.median(numeric_only=True))
```

```python
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nmissing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:",df.isnull().sum().sum())
```

```python
print("\nDuplicated rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicat rows:",df.duplicated().sum())
print("\nEDA process completed sucessfully!")
```
